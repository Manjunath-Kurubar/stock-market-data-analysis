```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(data['Date'],data['Close'])
plt.title('TCS Closing Price')
plt.xlabel('Date')
plt.ylabel('Price')
```

```python
data['MA7']=data['Close'].rolling(window=7).mean()
data['MA30']=data['Close'].rolling(window=30).mean()
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'], label='Close Price')
plt.plot(data['Date'], data['MA7'], label='7-Day MA')
plt.plot(data['Date'], data['MA30'], label='30-Day MA')
plt.title('TCS Price with Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
reliance = yf.download('RELIANCE.NS', start='2023-01-01', end='2024-12-31')
```

```python
reliance.reset_index(inplace=True)
```

```python
infy = yf.download('INFY.NS', start='2023-01-01', end='2024-12-31')
infy.reset_index(inplace=True)
print(infy.shape)
infy.head()
```

```python
reliance.head()
```

```python
infy['MA7'] = infy['Close'].rolling(window=7).mean()
infy['MA30'] = infy['Close'].rolling(window=30).mean()
```

```python
reliance['MA7'] = reliance['Close'].rolling(window=7).mean()
reliance['MA30'] = reliance['Close'].rolling(window=30).mean()
```

```python
reliance[['Close','MA7','MA30']].head(35)
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'], label='TCS')
plt.plot(infy['Date'], infy['Close'], label='INFY')
plt.plot(reliance['Date'], reliance['Close'], label='RELIANCE')
plt.title('Stock Price Comparison')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
data['Daily Return']=data['Close'].pct_change()
infy['Daily Return']=infy['Close'].pct_change()
reliance['Daily Return']=reliance['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
data['Daily Return']=data['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
print(data.columns)
print(data[('Close', 'TCS.NS')].dtype)
print(data[('Close', 'TCS.NS')].head(10))
```

```python
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
print(data.columns)
```

```python
data['Daily Return'] = data['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
infy.columns = [col[0] if isinstance(col, tuple) else col for col in infy.columns]
reliance.columns = [col[0] if isinstance(col, tuple) else col for col in reliance.columns]
print(infy.columns)
print(reliance.columns)
```

```python
infy['Daily Return'] = infy['Close'].pct_change()
reliance['Daily Return'] = reliance['Close'].pct_change()
display(infy[['Close','Daily Return']].head(10))
display(reliance[['Close','Daily Return']].head(10))
```

```python
data['Volatility'] = data['Daily Return'].rolling(window=7).std()
infy['Volatility'] = infy['Daily Return'].rolling(window=7).std()
reliance['Volatility'] = reliance['Daily Return'].rolling(window=7).std()
display(data[['Daily Return','Volatility']].head(10))
```

```python
print('TCS average daily return:', data['Daily Return'].mean())
print('INFY average daily return:', infy['Daily Return'].mean())
print('RELIANCE average daily return:', reliance['Daily Return'].mean())
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'],data['Close'],label='TCS')
plt.plot(infy['Date'], infy['Close'], label='INFY')
plt.plot(reliance['Date'], reliance['Close'], label='RELIANCE')
plt.title('Closing Price Comparison')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
data['Normalized'] = data['Close'] / data['Close'].iloc[0] * 100
infy['Normalized'] = infy['Close'] / infy['Close'].iloc[0] * 100
reliance['Normalized'] = reliance['Close'] / reliance['Close'].iloc[0] * 100
```

```python
print(data[['Close','Normalized']].head())
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Normalized'], label='TCS')
plt.plot(infy['Date'], infy['Normalized'], label='INFY')
plt.plot(reliance['Date'], reliance['Normalized'], label='RELIANCE')
plt.title('Normalized Growth Comparison (Base=100)')
plt.xlabel('Date')
plt.ylabel('Growth (%)')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(['TCS','INFY','RELIANCE'], [data['Volatility'].mean(), infy['Volatility'].mean(), reliance['Volatility'].mean()])
plt.title('Average Volatility Comparison')
plt.ylabel('Volatility')
plt.show()
```

## Insights for the growth chart:

Looking at the normalized growth chart, TCS showed the strongest and most consistent growth over the two-year period, generally staying ahead of INFY and RELIANCE. INFY had a rough start, dropping below its base value in early 2023, but recovered strongly and even matched TCS's growth by late 2024. RELIANCE stayed relatively stable in the middle range throughout, without major dips or spikes like the other two. By the end of the period, TCS and INFY had grown significantly more than RELIANCE.

## Insights for the volatility chart:
The volatility comparison shows INFY had the highest average volatility of the three stocks, meaning its daily price swings were slightly larger and less predictable. TCS and RELIANCE had similar, slightly lower volatility levels. Higher volatility isn't necessarily bad — it can mean more opportunity for short-term traders, but also more risk.

## Overall summary:

Overall, TCS delivered the strongest and most consistent growth with moderate volatility, making it the most stable performer in this analysis. INFY had higher volatility and a slower start but ultimately caught up in growth. RELIANCE was the most stable in terms of volatility but delivered the lowest returns of the three. This kind of comparison shows why looking at both returns and volatility together gives a more complete picture than price alone.
