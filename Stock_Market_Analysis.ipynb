```python
import yfinance as yf
import pandas as pd
import matplotlib as plt
```

```python
data=yf.download('TCS.NS',start='2023-01-01',end='2024-12-31')
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.reset_index(inplace=True)
```

```python
data.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(data['Date'],data['Close'])
plt.title('TCS Closing Price')
plt.xlabel('Date')
plt.ylabel('Price')
```

```python
data['MA7']=data['Close'].rolling(window=7).mean()
data['MA30']=data['Close'].rolling(window=30).mean()
```

```python
data[['Close','MA7','MA30']].head(35)
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'], label='Close Price')
plt.plot(data['Date'], data['MA7'], label='7-Day MA')
plt.plot(data['Date'], data['MA30'], label='30-Day MA')
plt.title('TCS Price with Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
data.isnull().sum()
```

```python
reliance = yf.download('RELIANCE.NS', start='2023-01-01', end='2024-12-31')
```

```python
reliance.reset_index(inplace=True)
```

```python
infy = yf.download('INFY.NS', start='2023-01-01', end='2024-12-31')
infy.reset_index(inplace=True)
print(infy.shape)
infy.head()
```

```python
reliance.head()
```

```python
infy['MA7'] = infy['Close'].rolling(window=7).mean()
infy['MA30'] = infy['Close'].rolling(window=30).mean()
```

```python
infy[['Close','MA7','MA30']].head(35)
```

```python
reliance['MA7'] = reliance['Close'].rolling(window=7).mean()
reliance['MA30'] = reliance['Close'].rolling(window=30).mean()
```

```python
reliance[['Close','MA7','MA30']].head(35)
```

```python
plt.figure(figsize=(12,6))
plt.plot(data['Date'], data['Close'], label='TCS')
plt.plot(infy['Date'], infy['Close'], label='INFY')
plt.plot(reliance['Date'], reliance['Close'], label='RELIANCE')
plt.title('Stock Price Comparison')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
data['Daily Return']=data['Close'].pct_change()
infy['Daily Return']=infy['Close'].pct_change()
reliance['Daily Return']=reliance['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
data['Daily Return']=data['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
print(data.columns)
print(data[('Close', 'TCS.NS')].dtype)
print(data[('Close', 'TCS.NS')].head(10))
```

```python
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
print(data.columns)
```

```python
data['Daily Return'] = data['Close'].pct_change()
display(data[['Close','Daily Return']].head(10))
```

```python
infy.columns = [col[0] if isinstance(col, tuple) else col for col in infy.columns]
reliance.columns = [col[0] if isinstance(col, tuple) else col for col in reliance.columns]
print(infy.columns)
print(reliance.columns)
```

```python
infy['Daily Return'] = infy['Close'].pct_change()
reliance['Daily Return'] = reliance['Close'].pct_change()
display(infy[['Close','Daily Return']].head(10))
display(reliance[['Close','Daily Return']].head(10))
```

```python
data['Volatility'] = data['Daily Return'].rolling(window=7).std()
infy['Volatility'] = infy['Daily Return'].rolling(window=7).std()
reliance['Volatility'] = reliance['Daily Return'].rolling(window=7).std()
display(data[['Daily Return','Volatility']].head(10))
```

```python
print('TCS average daily return:', data['Daily Return'].mean())
print('INFY average daily return:', infy['Daily Return'].mean())
print('RELIANCE average daily return:', reliance['Daily Return'].mean())
```
